# create dataloader

```python
label_b0 = df_ready0.label_id.values.reshape(-1,1)
accel_np0 = data_np_dict_b0['accel']
gyro_np0 = data_np_dict_b0['gyro']
euler_np0 = data_np_dict_b0['euler']
accdf_b0 = pd.DataFrame(np.concatenate([accel_np0, label_b0], axis=1), columns=['d1','d2','d3','label_id'])
gyrodf_b0 = pd.DataFrame(np.concatenate([gyro_np0, label_b0], axis=1), columns=['d1','d2','d3','label_id'])
eulerdf_b0 = pd.DataFrame(np.concatenate([euler_np0, label_b0], axis=1), columns=['d1','d2','d3','label_id'])

label_b1 = df_ready1.label_id.values.reshape(-1,1)
accel_np1 = data_np_dict_b1['accel']
gyro_np1 = data_np_dict_b1['gyro']
euler_np1 = data_np_dict_b1['euler']
accdf_b1 = pd.DataFrame(np.concatenate([accel_np1, label_b1], axis=1), columns=['d1','d2','d3','label_id'])
gyrodf_b1 = pd.DataFrame(np.concatenate([gyro_np1, label_b1], axis=1), columns=['d1','d2','d3','label_id'])
eulerdf_b1 = pd.DataFrame(np.concatenate([euler_np1, label_b1], axis=1), columns=['d1','d2','d3','label_id'])
```

```python
# prepare train loader of accelerometer
sensor_type = 'accel'
batch_size = 512
device = 'cuda'
len_sw = 180

df_ready = pd.concat([accdf_b0, accdf_b1], axis=0)
selected_columns = ['d1', 'd2', 'd3', 
                   'label_id']  # without timestamps
tmp_b = sliding_window(df_ready[selected_columns], len_sw)

# concatenate list
data_b = tmp_b[:,:,:-1]  # [B, Len, dim-1]
label_b = tmp_b[:,:,-1]  # [B, Len]
train_set_r = data_loader_umineko(data_b, label_b, device=device)
train_loader = DataLoader(train_set_r, batch_size=batch_size,
                           shuffle=False, drop_last=False)
```

```python
# separate back 0 and 1 to create testloaders
selected_columns = ['d1', 'd2', 'd3', 
                   'label_id']  # without timestamps
tmp_b = sliding_window(accdf_b0[selected_columns], len_sw)  # change here
# concatenate list
data_b = tmp_b[:,:,:-1]  # [B, Len, dim-1]
label_b = tmp_b[:,:,-1]  # [B, Len]
train_set_r = data_loader_umineko(data_b, label_b, device=device)
train_loader0 = DataLoader(train_set_r, batch_size=batch_size,
                           shuffle=False, drop_last=False)

tmp_b = sliding_window(accdf_b1[selected_columns], len_sw)  # change here
# concatenate list
data_b = tmp_b[:,:,:-1]  # [B, Len, dim-1]
label_b = tmp_b[:,:,-1]  # [B, Len]
train_set_r = data_loader_umineko(data_b, label_b, device=device)
train_loader1 = DataLoader(train_set_r, batch_size=batch_size,
                           shuffle=False, drop_last=False)
```

```python
from ae_model import Act2Vec

# Criterion
criterion = MSEloss()
criterion = criterion.to(device)

out_channels = 32
model = Act2Vec(out_channels, input_dim=(4503, 180, 3))
device = 'cuda'
model = model.to(device)

# Optimizer and scheduler
p_opti = 'sgd'
optimizer = get_optimizer(p_opti, model)
```

```python
# training
start_epoch = 0
num_epochs = 5000
lr = 0.0001
# i = 1  # 应该一只鸟一个trainloader，暂时全拼接到一起
# print('Starting %s-th bird data' % str(i))
for epoch in tqdm(range(start_epoch, num_epochs)):

    # Adjust lr
    lr_pre = lr
    lr = adjust_learning_rate(lr, optimizer, epoch, p_scheduler='cosine', p_epochs=num_epochs)
    # if (lr != lr_pre):
    #     print('Adjusted learning rate to {:.5f}'.format(lr))

    # Train: the same as simclr
    # print('Train ...')
    losses = AE_train_time_series(train_loader, model, criterion, optimizer)
    if (epoch % 1000 == 0) or (epoch == num_epochs-1):
        print('loss of the ' + str(epoch) + '-th training epoch is :' + losses.__str__())
        print('Saving model at: ' + 'AEgru_world_epoch%s' % str(epoch)\
      + '_datalen%s_' % str(len_sw) + sensor_type + '.pth')
torch.save(model.state_dict(), 'AEgru_world_epoch%s' % str(epoch) +\
           '_datalen%s_' % str(len_sw) + sensor_type + '.pth')

    # print('loss of the ' + str(epoch) + '-th training epoch is :' + losses.__str__())


# Save final model, xx\aaa-bbb-2024-04-24\unsup-models\iteration-0\aaaApr24\train
# column_list = '-'.join(data_column).replace('_','')
print('Saving model at: ' + 'AEgru_world_epoch%s' % str(epoch)\
      + '_datalen%s_' % str(len_sw) + sensor_type + '.pth')
torch.save(model.state_dict(), 'AEgru_world_epoch%s' % str(epoch) +\
           '_datalen%s_' % str(len_sw) + sensor_type + '.pth')

```
